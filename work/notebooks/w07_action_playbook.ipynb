## 1. Ranked actions + reason codes

#  Ranked Actions + Reason Codes

The content action playbook converts model observations into a ranked recommendation queue.

Each action includes:
- Priority level
- Recommended action
- Reason code
- Supporting signal

The ranking is designed for human decision-support. It does not automatically make publishing or business decisions.

## Action Mapping

| Signal | Recommended Action | Reason Code |
|---|---|---|
| Rising trend | Refresh and expand successful content | GROWTH_SIGNAL |
| Declining trend | Review and update content quality | DECAY_SIGNAL |
| Stable trend | Monitor performance over time | STABLE_MONITOR |


```python
import pandas as pd
df = pd.read_csv("/content/content_refresh_anonymized.csv")
# Create action queue from validated dataset
action_queue = df.copy()

def recommend_action(row):

    if str(row["trend_direction"]).lower() == "up":
        return "Refresh content and improve internal linking"

    elif str(row["trend_direction"]).lower() == "down":
        return "Review content quality and update information"

    else:
        return "Monitor content performance"


def generate_reason(row):

    if str(row["trend_direction"]).lower() == "up":
        return "GROWTH_SIGNAL"

    elif str(row["trend_direction"]).lower() == "down":
        return "DECAY_SIGNAL"

    else:
        return "STABLE_MONITOR"


action_queue["recommended_action"] = action_queue.apply(
    recommend_action,
    axis=1
)

action_queue["reason_code"] = action_queue.apply(
    generate_reason,
    axis=1
)


ranked_actions = action_queue[
    [
        "content_id",
        "client_id",
        "trend_direction",
        "recommended_action",
        "reason_code"
    ]
]


ranked_actions.head()
```

## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

#  Intended Use and Limits

## Intended Use

This playbook is designed for content teams to prioritize pages that may require attention.

Possible uses:
- Identify content needing refresh
- Prioritize review workload
- Support SEO/content planning decisions
- Monitor directional changes in content performance


## Limits

The recommendations are decision-support only.

The system does not:
- Guarantee future traffic growth
- Automatically publish or modify content
- Replace human expertise
- Make final business decisions
- Represent performance on every future dataset

Human review is required before taking action.

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

#  Human Review + No-Go List

## Human Review Rules

Before applying any recommendation:

1. A content specialist reviews the suggested action.
2. Business goals and audience context are considered.
3. Proposed changes are evaluated before publishing.
4. Results are monitored after changes.


## No-Go Cases

The system should NOT automatically:

- Delete content
- Publish content changes
- Change client-facing information
- Make business decisions
- Override expert review
- Treat predictions as guaranteed outcomes

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

# 4. Monitoring / Retrain Triggers

The recommendation system should be reviewed when:

## Data Drift

Possible triggers:
- Feature distributions change significantly
- New content categories appear
- User behavior patterns change


## Performance Drift

Possible triggers:
- Model accuracy decreases
- Recommendations become less useful
- Human feedback shows lower agreement


## Retraining Triggers

Retraining should be considered when:
- New validated data becomes available
- Performance decreases over time
- Business requirements change
